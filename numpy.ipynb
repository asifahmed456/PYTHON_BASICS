```python
import numpy as np
arr7= np.array([1,2,3,4,5,6])
reshaped= arr7.reshape(2,3)
print(reshaped)
reshaped2= reshaped.reshape(3,2)
print(reshaped2)
```

```python
ravel = reshaped2.ravel()
ravel[0]= 29
print(ravel)
print(reshaped2)
```

```python
flat= reshaped2.flatten()
flat[0]=10
print(flat)
print(reshaped)
```
